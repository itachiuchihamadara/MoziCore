```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

def plot_gaussian_distribution(data):
    """
    Plots a Gaussian distribution fitting the provided data.
    
    Args:
        data (list): List of integers to fit the distribution to
    """
    if not data:
        print("Error: Empty data list provided.")
        return
    
    # Convert data to numpy array for easier calculations
    data_array = np.array(data)
    
    # Calculate mean and standard deviation
    mu = np.mean(data_array)
    sigma = np.std(data_array)
    
    # Create a range of values covering the data range
    min_val = min(data)
    max_val = max(data)
    x = np.linspace(min_val - 3*sigma, max_val + 3*sigma, 1000)
    
    # Calculate the normal distribution curve
    y = norm.pdf(x, mu, sigma)
    
    # Create the plot
    plt.figure(figsize=(10, 6))
    
    # Plot the histogram of the actual data
    plt.hist(data, bins='auto', density=True, alpha=0.7, 
             color='skyblue', edgecolor='black', label='Data Histogram')
    
    # Plot the Gaussian curve
    plt.plot(x, y, 'r-', linewidth=2, label=f'Gaussian Fit (μ={mu:.2f}, σ={sigma:.2f})')
    
    # Add labels and title
    plt.title('Gaussian Distribution Fit to Data', fontsize=14)
    plt.xlabel('Value', fontsize=12)
    plt.ylabel('Probability Density', fontsize=12)
    plt.legend(fontsize=12)
    plt.grid(True, alpha=0.3)
    
    # Show the plot
    plt.tight_layout()
    plt.show()
```
